```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Code: Setting our success threshold based on Week 1's baseline
baseline_precision_at_50 = 0.240
target_precision_at_50 = 0.740

print(f"Baseline rule Precision@50: {baseline_precision_at_50}")
print(f"Our ML Goal Precision@50: {target_precision_at_50}")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed if-statement (like "IF impressions dropped by 20% THEN refresh") is too rigid. It falsely flags seasonal traffic drops and misses slow, steady performance bleeds. ML beats a fixed rule here because it can weigh non-linear interactions—such as how content age, current position, and keyword volume all interact together—without turning into an unmaintainable mess of nested if-statements.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Code: Proving data messiness (a simple rule fails because correlations are weak)
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search volume and actual impressions: {corr:.3f}")
print("Result: A fixed rule based purely on high search volume would fail to predict traffic.")
```
